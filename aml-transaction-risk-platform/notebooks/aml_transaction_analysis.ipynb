# AML Transaction Risk Analysis

Synthetic portfolio dataset for transaction monitoring and customer risk analysis.

```
import pandas as pd
from pathlib import Path
ROOT = Path('..')
```

```
tx = pd.read_csv(ROOT/'data/processed/transactions_clean.csv', parse_dates=['transaction_datetime'])
alerts = pd.read_csv(ROOT/'data/processed/suspicious_alerts.csv')
customers = pd.read_csv(ROOT/'data/processed/customer_risk.csv')
print('Transactions:', len(tx))
print('Alerts:', len(alerts))
print('Customers:', len(customers))
```

```
alerts.groupby('risk_tier').agg(alert_count=('alert_id','count'), total_value=('amount','sum')).reset_index()
```

```
customers['customer_risk_tier'].value_counts().rename_axis('risk_tier').reset_index(name='customers')
```
